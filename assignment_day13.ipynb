```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Load Dataset
df = pd.read_csv('Day13_Restaurant_Branch_Performance_Dataset.csv')

# 2. Store Type Summary Analysis
store_summary = df.groupby('Store_Type').agg(
    Branch_Count=('Branch', 'nunique'),
    Record_Count=('Record_ID', 'count'),
    Avg_Revenue=('Revenue', 'mean'),
    Total_Revenue=('Revenue', 'sum'),
    Avg_Profit=('Profit', 'mean'),
    Total_Profit=('Profit', 'sum'),
    Avg_Rating=('Customer_Rating', 'mean'),
    Avg_Customers=('Customers', 'mean')
).reset_index()

print("--- Store Type Performance Summary ---")
print(store_summary.to_string(index=False))

# 3. Regional Analysis
region_summary = df.groupby('Region').agg(
    Record_Count=('Record_ID', 'count'),
    Total_Revenue=('Revenue', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Avg_Rating=('Customer_Rating', 'mean'),
    Avg_Delivery=('Avg_Delivery_Min', 'mean')
).reset_index()

print("\n--- Region Performance Summary ---")
print(region_summary.to_string(index=False))

# 4. Correlation Matrix Calculation
numeric_cols = ['Staff_Count', 'Marketing_Spend', 'Customers', 'Orders', 
                'Average_Bill', 'Revenue', 'Food_Cost', 'Operating_Cost', 
                'Profit', 'Avg_Delivery_Min', 'Customer_Rating']
corr_matrix = df[numeric_cols].corr()

print("\n--- Correlation with Profit ---")
print(corr_matrix['Profit'].sort_values(ascending=False))

# 5. Optional Visualization: Correlation Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm', linewidths=0.5)
plt.title('Restaurant Branch Performance Correlation Matrix')
plt.tight_layout()
plt.savefig('correlation_matrix.png')
plt.show()
```
